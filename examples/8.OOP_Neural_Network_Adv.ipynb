Neural networks were designed to mimic the processing inside the human brain. They consist of a series of interconnected nodes (artificial neurones). Each neurone receives input signals, combines them using its weightings and bias, and can then pass a signal on to connected nodes.

In the diagram, signal strength between nodes with the strongest weightings are thicker representing a higher priority in determining the final output. The 'Bias' and 'Weightings' are learnt during the training cycle and used for predictions in the execution cycle.

#### Step 1: Import Dependencies

```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '1' 
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print(tf.__version__)
```

#### Step 2: Import And Visualise The Data

The MNIST dataset contains images of handwritten digits (0, 1, 2, etc.)

<img src="https://upload.wikimedia.org/wikipedia/commons/b/b1/MNIST_dataset_example.png"
         alt="MNIST sprite sample">

We will import the dataset from the Keras datasets API, then 60,000 images will be used to train the network and 10,000 images to evaluate how accurately the network learned to classify images.

```python
# Import images from the Keras datasets API into local variables
mnist = tf.keras.datasets.mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

Loading the dataset returns four NumPy arrays:

* The `train_images` and `train_labels` arrays are the *training set*—the data the model uses to learn.
* The model is tested against the *test set*, the `test_images`, and `test_labels` arrays.

The images are 28x28 NumPy arrays, with pixel values ranging from 0 to 255. The *labels* are an array of integers, ranging from 0 to 9. These correspond to the *class* of number the image represents.

Each image is mapped to a single label. Since the *class names* are not included with the dataset, store them here to use later when plotting the images:

```python
class_names = ['Class: 0', 'Class: 1', 'Class: 2', 'Class: 3', 'Class: 4',
               'Class: 5', 'Class: 6', 'Class: 7', 'Class: 8', 'Class: 9']
```

If you inspect the first image in the training set, you will see that the pixel values fall in the range of 0 to 255:

```python
plt.figure()
plt.imshow(train_images[0])
plt.title(f"Label: {train_labels[0]}")
plt.xlabel("28 Pixels")
plt.ylabel("28 Pixels")
plt.colorbar()
plt.grid(False)
plt.show()
```

As you can see, the image `train_images[0]` is `class 5`. The pixels that are dark purple (white if you looked at the handwritten digit) have a value of `0`. The pixels that are yellow (black if you looked at the handwritten digit) have a value of `255`.

#### Step 3: Wrangle The Data
The data must be wrangled before training the network to improve the model performance. We will scale these values to a range of 0 to 1 before feeding them to the neural network model. To do so, divide the values by 255. It's important that the *training set* and the *testing set* be wrangled in the same way:

```python
# Only scale the raw uint8 pixels, so re-running this cell does not divide by 255 twice
if train_images.dtype == np.uint8:
    train_images = train_images / 255.0
    test_images = test_images / 255.0

print("train_images range:", (train_images.min(), train_images.max()))
```

To verify that the data is in the correct format and that you're ready to build and train the network, let's visualise the first 25 images from `train_images` and display their class name below each image.

```python
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.xlabel(class_names[train_labels[i]])
plt.show()
```

#### Step 4: Instantiate a Neural Network Object And Configure The Layers

The basic building block of a neural network is the [*layer*](https://www.tensorflow.org/api_docs/python/tf/keras/layers). Layers extract representations from the data fed into them. Hopefully, these representations are meaningful for the problem at hand.

Most of deep learning consists of chaining together simple layers. Most layers, such as `tf.keras.layers.Dense`, have parameters that are learned during training.

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28)), # Input layer: each image is 28 x 28 pixels
    tf.keras.layers.Flatten(), # Flatten each image into 784 inputs
    tf.keras.layers.Dense(128, activation='relu'), # Hidden layer with 128 neurons and ReLU activation
    tf.keras.layers.Dense(10) # Output layer with 10 neurons for classification of 0-9
])
```

`tf.keras.Input` tells the model the shape of each image. The first layer in this network, `tf.keras.layers.Flatten`, transforms the format of the images from a two-dimensional array (of 28 by 28 pixels) to a one-dimensional array (of 28 * 28 = 784 pixels). Think of this layer as unstacking rows of pixels in the image and lining them up. This layer has no parameters to learn; it only reformats the data.

After the pixels are flattened, the network consists of a sequence of two `tf.keras.layers.Dense` layers. These are densely connected, or fully connected, neural layers. The first `Dense` layer has 128 nodes (or neurons). The second (and last) layer returns a logits array with length of 10. Each node contains a score that indicates the current image belongs to one of the 10 classes.

##### Compile the model

Before the model is ready for training, it needs a few more settings. These are added during the model's [*compile*](https://www.tensorflow.org/api_docs/python/tf/keras/Model#compile) step:

* [*Optimizer*](https://www.tensorflow.org/api_docs/python/tf/keras/optimizers) —This is how the model is updated based on the data it sees and its loss function.
* [*Loss function*](https://www.tensorflow.org/api_docs/python/tf/keras/losses) —This measures how accurate the model is during training. You want to minimize this function to "steer" the model in the right direction.
* [*Metrics*](https://www.tensorflow.org/api_docs/python/tf/keras/metrics) —Used to monitor the training and testing steps. The following example uses *accuracy*, the fraction of the images that are correctly classified.

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

#### Step 5: Training Cycle

Training Cycle (Fitting) of the neural network model requires the following steps:

1. Fit the training data to the model. In this example, the training data is in the `train_images` and `train_labels` arrays.
2. The model learns to associate images and labels through Training Cycles ($Epoch$). In each epoch, forward propagation makes predictions and backward propagation (backpropagation) adjusts the weightings and biases to reduce the loss.
3. Repetition of Training Cycles ($Epoch$) to improve accuracy and pattern matching.

```python
model.fit(train_images, train_labels, epochs=10)
```

#### Step 6: Evaluate The Model

```python
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)

print('\nTest accuracy:', test_acc)
```

#### Step 7: Execution Cycle

With the model fitted in the training cycle, you can use it to make predictions about some images. The model uses the weightings and biases learnt during the training cycle to determine the output.

In this example we have attached a softmax layer to convert the model's linear outputs ([logits](https://developers.google.com/machine-learning/glossary#logits)) to probabilities, which are easier to interpret.

```python
probability_model = tf.keras.Sequential([model,
                                         tf.keras.layers.Softmax()])
```

```python
predictions = probability_model.predict(test_images)
```

The execution cycle outputs a prediction as an array of 10 numbers. They represent the model's "confidence" that the image corresponds to each of the 10 different digits. You can see which label has the highest confidence value:

```python
print(predictions[0])
print(f"Predicted label: {np.argmax(predictions[0])}")
print(f"True label: {test_labels[0]}")
```

#### Step 8: Plot the Predictions

```python
def plot_image(i, predictions_array, true_label, img):
    true_label, img = true_label[i], img[i]
    plt.grid(False)
    plt.xticks([])
    plt.yticks([])

    plt.imshow(img, cmap=plt.cm.binary)

    predicted_label = np.argmax(predictions_array)
    if predicted_label == true_label:
        color = 'blue'
    else:
        color = 'red'

    plt.xlabel("{} {:2.0f}% ({})".format(class_names[predicted_label],
                                    100*np.max(predictions_array),
                                    class_names[true_label]),
                                    color=color)

def plot_value_array(i, predictions_array, true_label):
    true_label = true_label[i]
    plt.grid(False)
    plt.xticks(range(10))
    plt.yticks([])
    thisplot = plt.bar(range(10), predictions_array, color="#777777")
    plt.ylim([0, 1])
    predicted_label = np.argmax(predictions_array)

    thisplot[predicted_label].set_color('red')
    thisplot[true_label].set_color('blue')

i = 0
plt.figure(figsize=(6,3))
plt.subplot(1,2,1)
plot_image(i, predictions[i], test_labels, test_images)
plt.subplot(1,2,2)
plot_value_array(i, predictions[i],  test_labels)
plt.show()
```

Plot several images with their predictions. Note that the model can be wrong even when very confident.

```python
# Plot the first X test images, their predicted labels, and the true labels.
# Color correct predictions in blue and incorrect predictions in red.
num_rows = 5
num_cols = 3
num_images = num_rows*num_cols
plt.figure(figsize=(2*2*num_cols, 2*num_rows))
for i in range(num_images):
    plt.subplot(num_rows, 2*num_cols, 2*i+1)
    plot_image(i, predictions[i], test_labels, test_images)
    plt.subplot(num_rows, 2*num_cols, 2*i+2)
    plot_value_array(i, predictions[i], test_labels)
plt.tight_layout()
plt.show()
```

#### Step 9: Add Your Own Handwritten Numbers
Use this helper to load a photo or scan of your digit, convert it to MNIST style (28×28, grayscale, white digit on a black background) and predict.

Save an image such as **`my_digit.png`** in the same folder as this notebook, then run the cell below. Set `use_cnn=True` only after you have run the CNN section at the end of the notebook.

```python
from PIL import Image, ImageOps

# Image.Resampling was added in Pillow 9.1; older versions use Image.LANCZOS
LANCZOS = getattr(Image, "Resampling", Image).LANCZOS


def preprocess_digit(path, target_size=28, digit_size=20, show_steps=True):
    """Convert a photo or scan of one digit into MNIST format (28x28, white digit on black, values 0-1)."""
    original = Image.open(path).convert("L")
    img = original

    # MNIST digits are white on black, so invert images that have a light background
    if np.array(img).mean() > 127:
        img = ImageOps.invert(img)

    # Crop to the digit
    arr = np.array(img)
    mask = arr > 20
    if mask.any():
        coords = np.argwhere(mask)
        y0, x0 = coords.min(axis=0)
        y1, x1 = coords.max(axis=0) + 1
        img = Image.fromarray(arr[y0:y1, x0:x1])

    # Pad to a square so the digit is not stretched
    w, h = img.size
    side = max(w, h)
    square = Image.new("L", (side, side), color=0)
    square.paste(img, ((side - w) // 2, (side - h) // 2))

    # MNIST digits fit inside a 20x20 box in the centre of a 28x28 image
    digit = square.resize((digit_size, digit_size), LANCZOS)
    img28 = Image.new("L", (target_size, target_size), color=0)
    offset = (target_size - digit_size) // 2
    img28.paste(digit, (offset, offset))
    arr28 = np.asarray(img28).astype("float32") / 255.0

    if show_steps:
        fig, axs = plt.subplots(1, 3, figsize=(6, 2))
        axs[0].imshow(original, cmap="gray")
        axs[0].set_title("Original")
        axs[1].imshow(square, cmap="gray")
        axs[1].set_title("Cropped")
        axs[2].imshow(arr28, cmap="gray")
        axs[2].set_title("28×28")
        for ax in axs:
            ax.axis("off")
        plt.tight_layout()
        plt.show()
    return arr28


def predict_custom_digit(path, use_cnn=False):
    # Use models that output probabilities: probability_model adds softmax to the logits from `model`,
    # and model_cnn already ends in a softmax layer
    model_name = 'model_cnn' if use_cnn else 'probability_model'
    if model_name not in globals():
        if use_cnn:
            print("The CNN has not been trained yet. Run the CNN section at the end of the notebook first.")
        else:
            print("The model has not been trained yet. Run Steps 4 to 7 first.")
        return None, None

    img28 = preprocess_digit(path, show_steps=True)
    if use_cnn:
        x = img28[None, ..., None]
        probs = model_cnn.predict(x, verbose=0)[0]
    else:
        x = img28[None, ...]
        probs = probability_model.predict(x, verbose=0)[0]
    pred = int(np.argmax(probs))
    confidence = float(probs[pred])
    print(f"Prediction: {pred} (confidence {confidence:.3f})")
    return pred, probs


# Example usage: change the file name to match your image
image_path = 'my_digit.png'
if os.path.exists(image_path):
    pred, probs = predict_custom_digit(image_path, use_cnn=False)
else:
    print(f"'{image_path}' not found. Save a photo of a handwritten digit in the same folder as this notebook, then run this cell again.")
```

#### Convolutional Neural Network (CNN)
CNNs are better for images. This is a minimal example.

```python
# Prepare data for the CNN: add a channel dimension, (28, 28) -> (28, 28, 1)
train_images_cnn = np.expand_dims(train_images, -1)
test_images_cnn = np.expand_dims(test_images, -1)

model_cnn = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28, 1)), # Input layer: 28 x 28 pixels with 1 colour channel (greyscale)
    tf.keras.layers.Conv2D(16, (3, 3), activation='relu'), # 16 filters that detect small patterns such as edges
    tf.keras.layers.MaxPooling2D((2, 2)), # Halve the width and height, keeping the strongest signals
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'), # 32 filters that combine edges into larger shapes
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax') # Softmax outputs probabilities directly
])

# The labels are integers (0-9), so use SparseCategoricalCrossentropy as in Step 4
model_cnn.compile(optimizer='adam',
                  loss=tf.keras.losses.SparseCategoricalCrossentropy(),
                  metrics=['accuracy'])

history_cnn = model_cnn.fit(train_images_cnn, train_labels, epochs=3, batch_size=32, validation_split=0.1, verbose=2)

test_loss_cnn, test_acc_cnn = model_cnn.evaluate(test_images_cnn, test_labels, verbose=0)
print("CNN Test accuracy:", round(float(test_acc_cnn), 4))
```
